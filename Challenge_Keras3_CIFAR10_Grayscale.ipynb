## 0. Librerías


````python
import keras
from keras import layers
import numpy as np
import matplotlib.pyplot as plt

SEED = 42
keras.utils.set_random_seed(SEED)

print(f"Keras version : {keras.__version__}")
print(f"Backend       : {keras.backend.backend()}")
````

# 1. Dataset: CIFAR-10

CIFAR-10 contiene 50,000 imágenes de entrenamiento y 10,000 de prueba. Las imágenes originales son de **32×32×3** y pertenecen a 10 categorías:

`airplane, automobile, bird, cat, deer, dog, frog, horse, ship, truck`.


````python
# Esta parte está resuelta.
(X_train, y_train), (X_test, y_test) = keras.datasets.cifar10.load_data()

y_train = y_train.squeeze()
y_test = y_test.squeeze()

class_names = [
    "airplane", "automobile", "bird", "cat", "deer",
    "dog", "frog", "horse", "ship", "truck"
]

print("X_train original:", X_train.shape)
print("X_test original :", X_test.shape)
print("y_train         :", y_train.shape)
````

## 1.1 RGB → escala de grises

Usaremos la transformación:

$$Gray = 0.299R + 0.587G + 0.114B$$

Esta parte está resuelta porque el objetivo del ejercicio es practicar Keras.


````python
X_train = (
    0.299 * X_train[..., 0] +
    0.587 * X_train[..., 1] +
    0.114 * X_train[..., 2]
)

X_test = (
    0.299 * X_test[..., 0] +
    0.587 * X_test[..., 1] +
    0.114 * X_test[..., 2]
)

print("X_train grayscale:", X_train.shape)
print("X_test grayscale :", X_test.shape)
````

### Pregunta 1

Después de convertir las imágenes a escala de grises, cada imagen tiene tamaño `32×32`.

¿Cuántos valores tendrá cada imagen después de aplicar `Flatten()`?

**Respuesta:** 1024 valores. `Flatten()` solo reacomoda la matriz 32×32 en un vector unidimensional, así que 32 × 32 = 1024. (Si no hubiéramos pasado a grises serían 32 × 32 × 3 = 3072.)

## 1.2 Normalización

Los valores de los píxeles están entre 0 y 255. Completa el código para llevarlos al rango `[0,1]`.


````python
# TODO 1 — Normalización
X_train = X_train.astype("float32") / 255.0
X_test  = X_test.astype("float32") / 255.0

print(X_train.min(), X_train.max())
````

````python
# Validación TODO 1
assert X_train.dtype == np.float32
assert X_test.dtype == np.float32
assert X_train.min() >= 0.0
assert X_train.max() <= 1.0
print("✓ Datos normalizados correctamente")
````

## 1.3 Visualización

Esta parte está resuelta.


````python
plt.figure(figsize=(12, 6))
for i in range(12):
    plt.subplot(3, 4, i + 1)
    plt.imshow(X_train[i], cmap="gray")
    plt.title(class_names[y_train[i]])
    plt.xticks([])
    plt.yticks([])
plt.tight_layout()
plt.show()
````

---
# 2. Construir la red neuronal

Construiremos la misma arquitectura conceptual vista en clase:

```text
Imagen 32×32
   ↓
Flatten
   ↓
1024 valores
   ↓
Dense(128)
   ↓
ReLU
   ↓
Dense(10)
   ↓
Logits
```

> La última capa **no** debe tener Softmax. Trabajaremos con logits.


````python
# TODO 2 — Completa la arquitectura
model = keras.Sequential([
    keras.Input(shape=(32, 32)),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(10)
])

model.summary()
````

````python
# Validación TODO 2
expected_params = (1024 * 128 + 128) + (128 * 10 + 10)
assert model.count_params() == expected_params
print(f"✓ Arquitectura correcta: {model.count_params():,} parámetros")
````

### Preguntas

**2.** ¿Por qué necesitamos `Flatten()` antes de la primera capa `Dense`?

**Respuesta:** Porque una capa `Dense` espera como entrada un vector (1D) por ejemplo: conecta cada valor de entrada con cada neurona. La imagen llega como una matriz 32×32 (2D), así que `Flatten()` la convierte en un vector de 1024 valores para que se pueda hacer la multiplicación matricial. No aprende nada ni tiene parámetros: solo cambia la forma de los datos.

**3.** ¿Por qué la última capa tiene 10 neuronas?

**Respuesta:** Porque CIFAR-10 tiene exactamente 10 clases (airplane, automobile, bird, cat, deer, dog, frog, horse, ship, truck). Cada neurona de salida produce la puntuación (logit) de una clase, y la clase predicha es la del logit más alto.

**4.** ¿Qué función cumple `ReLU`?

**Respuesta:** Introduce no linealidad: `ReLU(x) = max(0, x)`. Sin una activación no lineal, apilar `Dense` tras `Dense` sería equivalente a una sola transformación lineal y la red no podría aprender relaciones complejas. Además es barata de calcular y evita en buena medida el problema del gradiente que se desvanece.

---
# 3. Configurar el entrenamiento

Usaremos:

- `Adam`
- `SparseCategoricalCrossentropy`
- `accuracy`

Como la red produce logits, la pérdida debe configurarse con `from_logits=True`.


````python
lr_rate = 0.001

# TODO 3 — Configura el entrenamiento
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=lr_rate),
    loss=keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=["accuracy"]
)
````

### Pregunta 5

¿Por qué usamos `from_logits=True`?

**Respuesta:** Porque la última capa no tiene `softmax`: entrega logits (números reales sin normalizar). Con `from_logits=True` le decimos a la función de pérdida que aplique ella misma el softmax internamente, usando una implementación numéricamente estable (log-sum-exp). Si pusiéramos softmax en la capa y además `from_logits=True`, estaríamos aplicando softmax dos veces y el entrenamiento se degradaría.

Usamos `SparseCategoricalCrossentropy` (y no la versión `Categorical`) porque las etiquetas son enteros (`0..9`) y no vectores one-hot.

---
# 4. Entrenar el modelo

Usaremos 10 épocas y mini-batches de 64 ejemplos. Completa `fit()`.


````python
epochs = 10
batch_size = 64

# TODO 4 — Entrenamiento
history = model.fit(
    X_train,
    y_train,
    epochs=epochs,
    batch_size=batch_size,
    validation_data=(X_test, y_test),
    shuffle=True
)
````

### Preguntas

**6.** ¿Qué representa una `epoch`?

**Respuesta:** Una pasada completa por todo el conjunto de entrenamiento (las 50,000 imágenes). Con 10 épocas, la red ve el dataset completo 10 veces y ajusta sus pesos en cada pasada.

**7.** ¿Qué representa `batch_size=64`?

**Respuesta:** El número de imágenes que se procesan juntas antes de actualizar los pesos una vez. Con 50,000 imágenes y lotes de 64 hay ≈782 actualizaciones por época. Lotes más grandes dan gradientes más estables y mejor uso de la GPU; lotes más pequeños dan más actualizaciones y más ruido (que a veces ayuda a generalizar).

## 4.1 Curvas de aprendizaje

Esta parte está resuelta.


````python
plt.figure(figsize=(7, 4))
plt.plot(history.history["accuracy"], label="Train accuracy")
plt.plot(history.history["val_accuracy"], label="Validation accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Evolución del entrenamiento")
plt.legend()
plt.show()
````

### Pregunta 8

Observa las dos curvas. ¿Hay evidencia de overfitting? Justifica brevemente.

**Respuesta:** _(ajusta esta respuesta con tus curvas reales)_ Sí, se observa overfitting a partir de las primeras épocas: el accuracy de entrenamiento sigue subiendo mientras el de validación se estanca (típicamente alrededor de 0.35–0.40) e incluso empieza a bajar, abriendo una brecha creciente entre ambas curvas. Eso indica que la red empieza a memorizar ejemplos concretos en lugar de aprender patrones generalizables. Se podría mitigar con `Dropout`, regularización L2, *data augmentation* o *early stopping*.

# 5. Evaluar el modelo

Completa `evaluate()`.


````python
# TODO 5 — Evaluación
test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    batch_size=batch_size,
    verbose=0
)

print(f"Test loss     : {test_loss:.4f}")
print(f"Test accuracy : {test_accuracy:.4f}")
print(f"Accuracy (%)  : {test_accuracy * 100:.2f}%")
````

# 6. Logits y Softmax

La salida de la red son 10 logits. Para interpretarlos como probabilidades necesitamos Softmax.


````python
index = 0
image = X_test[index]
true_class = y_test[index]

plt.imshow(image, cmap="gray")
plt.title(f"Clase real: {class_names[true_class]}")
plt.axis("off")
plt.show()

image_batch = np.expand_dims(image, axis=0)
logits = model(image_batch, training=False)

print("Logits:")
print(keras.ops.convert_to_numpy(logits))
````

````python
# TODO 6 — Logits -> probabilidades
probabilities = keras.ops.softmax(logits)

# TODO 7 — Clase con mayor probabilidad
prediction = keras.ops.argmax(probabilities, axis=1)
prediction = int(keras.ops.convert_to_numpy(prediction)[0])

print("Clase real     :", class_names[true_class])
print("Clase predicha :", class_names[prediction])
````

````python
probabilities_np = keras.ops.convert_to_numpy(probabilities)[0]

print("\nProbabilidades:")
for i, p in enumerate(probabilities_np):
    marker = " ← predicción" if i == prediction else ""
    print(f"{class_names[i]:12s}: {p:.4f}{marker}")

print("\nSuma:", probabilities_np.sum())
````

### Pregunta 10

¿Por qué las probabilidades producidas por `Softmax` deben sumar aproximadamente 1?

**Respuesta:** Porque softmax convierte los logits en una distribución de probabilidad sobre las 10 clases: exponencia cada logit y divide por la suma de todas las exponenciales, así que por construcción cada valor queda en (0,1) y el total suma 1. Conceptualmente, el modelo debe repartir el 100% de su "creencia" entre clases mutuamente excluyentes: la imagen es exactamente una de las diez. La suma puede dar 0.9999999 en lugar de 1 exacto por redondeo de punto flotante.

### 2. ¿Qué clases parecen más difíciles de distinguir?

### 3. Después de `Flatten()`, la imagen se convierte en un vector de 1024 valores. ¿Qué información importante sobre la imagen podría perderse al tratarla únicamente como un vector?

**Respuesta:** Se pierde la vecindad espacial: para la capa `Dense`, el píxel (5,5) y el (5,6) son dos entradas independientes sin ninguna relación, aunque en la imagen estén pegados. Eso destruye la noción de bordes, texturas y formas locales. Tampoco hay invariancia a traslación: si el mismo objeto aparece desplazado unos píxeles, el vector cambia por completo y la red debe volver a aprenderlo desde cero.

### 4. ¿Qué tipo de arquitectura podría aprovechar mejor la estructura espacial de las imágenes?

**Respuesta:** Una **red neuronal convolucional (CNN)**. Aplica filtros pequeños que recorren la imagen detectando patrones locales (bordes, esquinas, texturas) y los combina jerárquicamente en formas y objetos. Comparte los pesos de cada filtro en toda la imagen, lo que da invariancia a traslación y muchísimos menos parámetros, y el *pooling* le da robustez a pequeños desplazamientos.
